# E-Commerce Sales Analysis Project
This project analyzes sales and profit trends using Python and prepares data for a Power BI dashboard.

```python
import pandas as pd
df = pd.read_csv('../data/superstore_sales.csv')
df.head()
```

```python
# Basic dataset info
df.info()
```

```python
# Convert date
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df.head()
```

```python
# Sales by Region
sales_region = df.groupby('Region')['Sales'].sum()
sales_region
```

```python
# Profit by Category
profit_category = df.groupby('Category')['Profit'].sum()
profit_category
```

```python
# Monthly sales trend
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales
```

```python
# Save cleaned dataset for dashboard
df.to_csv('../data/cleaned_superstore.csv', index=False)
```
