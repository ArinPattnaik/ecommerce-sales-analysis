# E-Commerce Sales Analysis Project

This notebook demonstrates comprehensive sales analysis using the Superstore dataset. The analysis includes data exploration, key insights, and visualizations.

## Project Overview
- **Goal**: Analyze e-commerce sales data to uncover trends, profitability patterns, and business insights
- **Dataset**: Superstore sales data with order details, customer segments, and product categories
- **Tools**: Python (Pandas, NumPy, Matplotlib, Seaborn, Plotly), Streamlit for dashboard

```python
# Import libraries
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Add src to path
sys.path.append('../src')

from data_loader import load_data, preprocess_data
from analysis import (
    sales_by_dimension, profit_by_dimension, monthly_trends,
    top_products, correlation_analysis, key_metrics, discount_impact
)
from visualization import (
    plot_sales_by_region, plot_profit_by_category, plot_monthly_trends,
    create_interactive_dashboard, plot_correlation_heatmap, plot_discount_impact
)

# Set plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

print("Libraries imported successfully!")
```

```python
# Load and preprocess data
data_path = '../data/superstore_sales.csv'
df = load_data(data_path)
df = preprocess_data(df)

print(f"Dataset shape: {df.shape}")
print(f"Date range: {df['Order Date'].min()} to {df['Order Date'].max()}")
df.head()
```

```python
# Dataset overview
print("Dataset Info:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nBasic statistics:")
df.describe()
```

```python
# Key Metrics
metrics = key_metrics(df)
print("Key Business Metrics:")
for key, value in metrics.items():
    if 'Sales' in key or 'Profit' in key:
        print(f"{key}: ${value:,.0f}")
    elif 'Margin' in key:
        print(f"{key}: {value:.1f}%")
    else:
        print(f"{key}: {value:,.0f}")
```

```python
# Sales Analysis
print("Sales by Region:")
sales_region = sales_by_dimension(df, 'Region')
print(sales_region)

print("\nSales by Category:")
sales_cat = sales_by_dimension(df, 'Category')
print(sales_cat)

print("\nSales by Segment:")
sales_seg = sales_by_dimension(df, 'Segment')
print(sales_seg)
```

```python
# Profit Analysis
print("Profit by Region:")
profit_region = profit_by_dimension(df, 'Region')
print(profit_region)

print("\nProfit by Category:")
profit_cat = profit_by_dimension(df, 'Category')
print(profit_cat)

print("\nProfit by Segment:")
profit_seg = profit_by_dimension(df, 'Segment')
print(profit_seg)
```

```python
# Monthly Trends
monthly_data = monthly_trends(df)
print("Monthly Sales and Profit Trends:")
monthly_data.head(12)
```

```python
# Top Products
top_10_products = top_products(df, 10)
print("Top 10 Products by Sales:")
top_10_products
```

```python
# Correlation Analysis
correlation_matrix = correlation_analysis(df)
print("Correlation Matrix:")
correlation_matrix
```

```python
# Discount Impact Analysis
discount_analysis = discount_impact(df)
print("Discount Impact on Sales and Profit:")
discount_analysis
```

```python
# Visualizations

# Sales by Region
fig1 = plot_sales_by_region(df)
plt.show()

# Profit by Category
fig2 = plot_profit_by_category(df)
plt.show()
```

```python
# Monthly Trends Plot
fig3 = plot_monthly_trends(df)
plt.show()
```

```python
# Correlation Heatmap
fig4 = plot_correlation_heatmap(df)
plt.show()
```

```python
# Discount Impact Plot
fig5 = plot_discount_impact(df)
plt.show()
```
